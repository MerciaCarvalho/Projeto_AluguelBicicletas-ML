# atividade previsao aluguel bicicletas

```python
from sklearn.linear_model import LinearRegression
```

```python

duracoes = [10, 15, 20, 25, 30, 35, 40, 50, 55, 60]
precos = [10, 13, 14, 18, 19, 23, 24, 30, 32, 34]

# Entrada: duração em minutos
X = [[x] for x in duracoes]

# Saída: preço em reais
y = precos

# Criar o modelo
modelo = LinearRegression()

# Treinar o modelo
modelo.fit(X, y)

# Previsões
previsao_45 = modelo.predict([[45]])
previsao_20 = modelo.predict([[20]])
previsao_60 = modelo.predict([[60]])

print(f"45 minutos: R$ {previsao_45[0]:.2f}")
print(f"20 minutos: R$ {previsao_20[0]:.2f}")
print(f"60 minutos: R$ {previsao_60[0]:.2f}")
```
